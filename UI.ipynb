### 6. Column

- 실제 Column 컴포저블 래퍼 함수 정의
```python
@Composable
def Column(modifier: Modifier,
            vertical_arrangement: Arrangement.Vertical = Arrangement.Center
             horizontal_alignment: Alignment.Horizontal = Alignment.Horizontal.Star
              content=lambda: {}): ...
```

````python
@Composable
def Practice():  # 세로 정렬
    Column(
        modifier=modifier,
        content=lambda: {
            Button(
                onclick=lambda: None,
                corner_radius=10,
                color=0xFF000000,
                content=lambda: Text("작은 버튼", color=0xFFFFFFFF, font_size=10)
            ),
            Button(
                onclick=lambda: None,
                corner_radius=10,
                color=0xFF000000,
                content=lambda: Text("큰 버튼", color=0xFFFFFFFF, font_size=30)
            )
        }
    )
main.App.update(Practice)
````

````python
@Composable
def Practice():  # 세로 정렬 + 오른쪽 정렬
    Column(
        modifier=modifier,
        horizontal_alignment=Alignment.Horizontal.End,
        content=lambda: {
            Text("작은 글자", color=0xFF000000, font_size=10),
            Text("큰 글자", color=0xFF000000, font_size=30)
        }
    )
main.App.update(Practice)
````

### 7. Row

- 실제 Row 컴포저블 래퍼 함수 정의
```python
@Composable
def Row(modifier: Modifier,
         horizontal_arrangement: Arrangement.Horizontal = Arrangement.Center
          vertical_alignment: Alignment.Vertical = Alignment.Vertical.To
           content=lambda: {}): ...
```

````python
@Composable
def Practice():  # 가로 정렬
    Row(
        modifier=modifier,
        content=lambda: {
            Text("작은 글자", color=0xFF000000, font_size=10),
            Text("큰 글자", color=0xFF000000, font_size=30)
        }
    )
main.App.update(Practice)
````

### 8. Spacer

- 실제 Spacer 컴포저블 래퍼 함수 정의
```python
@Composable
def Spacer(start: : float = 0.0, top: float = 0.0, end: float = 0.0, bottom: float = 0.0 : ...
```

````python
@Composable
def Practice():  # 아까 카드에 여백 넣어보기
    Card(
        modifier=modifier,
        corner_radius=20,
        color=0xFF000000,
        content=lambda: {
            Column(
                modifier=modifier,
                content=lambda: {
                    Spacer(top=10.0),
                    Row(
                        modifier=modifier,
                        content=lambda: {
                            Spacer(start=20.0),
                            Text("나는 그냥 텍스트랍니다.", color=0xFFFFFFFF),
                            Spacer(end=20.0)
                        }
                    ),
                    Spacer(bottom=10.0)
                }
            )
        }
    )
main.App.update(Practice)
````

## 온디바이스 AI 실습

##### Llama3.1 모델 동작 및 UI 반영

````python
# 모델 Import
from model import llama3 as _llama3
from model.config import ChatHistory
````

````python
token_streamer = _llama3.token_streamer
llama3 = _llama3.chat
chat_history = ChatHistory()
````

````python
# UI State 업데이트 함수들
def change_prompt(prompt: str):
    main.App.user_prompt.setValue(prompt)

def print_state(text: str):
    main.App.status.setValue(main.App.status.getValue() + "  " + text)

def print_messages(text: str):
    main.App.messages.setValue(main.App.messages.getValue() + text)
````

````python
@Composable
def LlamaView():  # 모델 동작 관련 UI 디자인
    Text(f"Current User Prompt:  {main.App.user_prompt.getValue()}")
    Text(f"Log:{main.App.status.getValue()}")
    Text("")
    Text(main.App.messages.getValue())
    Button(
        onclick=lambda: main.run_llama3(printer=print_messages),
        content=lambda: {
            Text(f"유저 프롬프트 전송")
        }
    )
    Button(
        onclick=lambda: {
            change_prompt("오늘 날씨는 어때요?")
        },
        content=lambda: {
            Text(f"프롬프트 변경")
        }
    )
````

````python
main.App.update(LlamaView)
````

````python
# 토큰 스트리밍
_user_prompt = main.App.user_prompt.getValue()
print_state("Inference...")

for chunk in token_streamer(*llama3(chat_history, _user_prompt)):
    print_messages(chunk)
print_messages("\n")

print_state("Done!")
````
